Q2. Data Cleaning

a) Check for missing values in the dataset.
b) Handle missing values appropriately.
c) Identify categorical columns present in the dataset.

```python
Python code

import pandas as pd

# a) Check for missing values
print("Missing Values:")
print(df.isnull().sum())

# b) Handle missing values appropriately

# Fill numerical columns with mean
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())

# Fill categorical columns with mode
cat_cols = df.select_dtypes(include=['object', 'category']).columns

for col in cat_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Check missing values again
print("Missing Values After Cleaning:")
print(df.isnull().sum())

# c) Identify categorical columns
print("Categorical Columns:")
print(cat_cols.tolist())

Explanation
1. df.isnull().sum(): Checks missing values in each column.
2. Mean: Fills missing numerical values with the average.
3. Mode: Fills missing categorical values with the most frequently occurring value.
4. select_dtypes(): Identifies categorical columns such as Name, City, Gender, etc.
Note: Run Q1 first, then Q2 in the same notebook so that df is already loaded.
```

Q4. Outlier Detection and Treatment

a) Detect outliers in numerical columns using the IQR method.
b) Treat outliers using capping techniques

```python
Python code

import pandas as pd
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# a) Detect outliers using IQR method

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"\nOutliers in {col}:")
    print("Number of Outliers:", len(outliers))


# b) Treat outliers using Capping Technique

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df[col] = np.clip(df[col], lower_limit, upper_limit)

print("\nOutliers treated successfully!")

Explanation

a) Outlier Detection
- Q1 = First Quartile (25%).
- Q3 = Third Quartile (75%).
- IQR = Q3 - Q1.
- Values below the lower limit or above the upper limit are considered outliers.

b) Outlier Treatment

- Uses the capping technique.
- Values below the lower limit are replaced with the lower limit.
- Values above the upper limit are replaced with the upper limit.
- This helps reduce the effect of extreme values.

Expected Output:

Outliers in Age:
Number of Outliers: 5

Outliers in Salary:
Number of Outliers: 8

Outliers treated successfully!
```

Q6. Feature Selection and Data Splitting

a) Separate independent variables (X) and dependent variable (y).
b) Split the dataset into training and testing sets.

```python
Python code

from sklearn.model_selection import train_test_split

# a) Separate independent variables (X) and dependent variable (y)

X = df.drop('Target', axis=1)
y = df['Target']

print("Independent Variables (X):")
print(X.head())

print("\nDependent Variable (y):")
print(y.head())


# b) Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Display dataset shapes

print("\nTraining and Testing Data:")

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

Explanation

a) Feature Selection
- X = Independent variables (features).
- y = Dependent variable (Target).

b) Data Splitting
- test_size=0.20 means 80% data for training and 20% for testing.
- random_state=42 ensures reproducible results.

Expected Output

X_train: (800, 5)
X_test: (200, 5)
y_train: (800,)
y_test: (200,)

Note: Output is an example. Actual values depend on your dataset.
```

Q7. Apply feature scaling to the dataset using StandardScaler.

```python
Python code

from sklearn.preprocessing import StandardScaler

# Create StandardScaler object
scaler = StandardScaler()

# Fit and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data
X_test_scaled = scaler.transform(X_test)

# Display scaled data
print("Scaled Training Data:")
print(X_train_scaled[:5])

print("\nScaled Testing Data:")
print(X_test_scaled[:5])

Explanation
1. StandardScaler(): Converts numerical features into a standard scale.
2. fit_transform(): Calculates mean and standard deviation from training data and scales it.
3. transform(): Applies the same scaling to testing data.
4. X_train_scaled: Stores scaled training data.
5. X_test_scaled: Stores scaled testing data.

StandardScaler Formula
\[
Z = \frac{X-\mu}{\sigma}
\]

Where:

- X = Original value
- μ = Mean
- σ = Standard deviation
- Z = Scaled value

Expected Output:

Scaled Training Data:
[[-0.45  0.67  1.23]
 [ 0.32 -0.56  0.78]
 [-1.12  0.45 -0.34]]

 Note: Output is only an example.
```

Q10. Model Evaluation – ROC Curve & AUC

a) Plot the ROC curve.
b) Calculate the AUC score.

```python
Python code

from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# a) Calculate prediction probabilities
y_prob = model.predict_proba(X_test_scaled)[:, 1]

# Generate ROC Curve values
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Plot ROC Curve
plt.figure(figsize=(8, 5))

plt.plot(fpr, tpr, label='Logistic Regression')
plt.plot([0, 1], [0, 1], linestyle='--', color='red')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()


# b) Calculate AUC Score
auc_score = roc_auc_score(y_test, y_prob)

print("AUC Score:", auc_score)

Explanation

a) ROC Curve
- ROC stands for Receiver Operating Characteristic.
- It shows the performance of a classification model at different classification thresholds.
- X-axis represents False Positive Rate.
- Y-axis represents True Positive Rate.

b) AUC Score
- AUC stands for Area Under the Curve.
- It measures how well the model separates two classes.

AUC Score Interpretation
AUC Score	Model Performance
0.90 – 1.00	Excellent
0.80 – 0.90	Very Good
0.70 – 0.80	Good
0.60 – 0.70	Average
0.50 – 0.60	Poor
0.50	Random prediction

ROC

Conclusion
A higher AUC score indicates better classification performance. An AUC closer to 1 means the model can distinguish between positive and negative classes more effectively.

Note: The actual AUC score depends on your dataset and trained model. This code assumes binary classification with target labels 0 and 1.
```
